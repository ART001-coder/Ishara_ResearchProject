# 03 — Final Model Evaluation & Confusion Matrix Analysis

**Reference:** `implementation_plan.md` → Section 10 (Success Criteria), Section 8.2 (Bug M5: Confused Sign Pairs), Section 9 (Day 9 Tasks 9.1–9.5)

**Purpose of this notebook:**
- Evaluate trained model on test set (`data/test_split.csv`).
- Compute Top-1 and Top-3 accuracy.
- Render and save confusion matrix heatmap (`logs/confusion_matrix.png`).
- Extract top-10 confused sign pairs.

---


```python
import os
import json
import torch
import numpy as np
import pandas as pd
from src.data.dataset import get_dataloaders
from src.model.lstm_model import ISLRecognizer
from src.utils.metrics import calculate_topk_accuracy, generate_confusion_matrix, find_top_confused_pairs
from src.utils.config import load_config

cfg = load_config('../config.yaml')
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Load test dataloader
_, _, test_loader = get_dataloaders('../config.yaml')

# Load model checkpoint
model = ISLRecognizer(num_classes=65).to(device)
checkpoint_path = '../checkpoints/best_model.pth'
if os.path.exists(checkpoint_path):
    model.load_state_dict(torch.load(checkpoint_path, map_location=device))
    print(f'[SUCCESS] Loaded test checkpoint: {checkpoint_path}')
model.eval()

y_true = []
y_logits = []
with torch.no_grad():
    for bx, by in test_loader:
        logits = model(bx.to(device))
        y_true.extend(by.numpy())
        y_logits.extend(logits.cpu().numpy())

y_true = np.array(y_true)
y_logits = np.array(y_logits)
y_pred = np.argmax(y_logits, axis=1)

# Compute Metrics
top1_acc = calculate_topk_accuracy(y_true, y_logits, k=1)
top3_acc = calculate_topk_accuracy(y_true, y_logits, k=3)
print(f'Top-1 Test Accuracy: {top1_acc*100:.2f}%')
print(f'Top-3 Test Accuracy: {top3_acc*100:.2f}%')

# Vocabulary list
with open('../data/vocabulary.json') as f:
    vocab = json.load(f)
classes = [vocab['id_to_word'][str(i)] for i in range(65)]

cm = generate_confusion_matrix(y_true, y_pred, classes, output_path='../logs/confusion_matrix.png')
confused_pairs = find_top_confused_pairs(cm, classes, top_n=10)
print('Top 10 Confused Sign Pairs:', confused_pairs)
```
